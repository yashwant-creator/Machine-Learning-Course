```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv("Data.csv")
X = dataset.iloc[:,:-1].values
y = dataset.iloc[:,-1].values
```

```python
print(X)
```

```python
print(y)
```

```python
# Splitting the training and test set
from sklearn.model_selection import train_test_split
X_train,X_test, y_train, y_test = train_test_split(X,y, test_size = 0.2, random_state = 0)
```

```python
# applying multiple linear regression model on the training set
from sklearn.linear_model import LinearRegression
multi_reg = LinearRegression()
multi_reg.fit(X_train,y_train)
```

```python
#predicting the test results
y_pred = multi_reg.predict(X_test)
```

```python
plt.scatter(X_test[:,0], y_test, color = 'blue')
plt.scatter(X_test[:,0], y_pred, color = 'red')
plt.title("Engine temp vs energy produced")
```

```python
plt.scatter(X_test[:,1], y_test, color = 'blue')
plt.scatter(X_test[:,1], y_pred, color = 'red')
plt.title("Engine volume vs energy produced")
```

```python
plt.scatter(X_test[:,2], y_test, color = 'blue')
plt.scatter(X_test[:,2], y_pred, color = 'red')
plt.title("Engine pressure vs energy produced")
```

```python
plt.scatter(X_test[:,0], y_test, color = 'blue')
plt.scatter(X_test[:,0], y_pred, color = 'red')
plt.title("Engine RH vs energy produced")
```

```python
# evaluating the model performance
from sklearn.metrics import r2_score
print("Multi-Linear Regression R^2")
r2_score(y_test, y_pred)
```

```python
# Polynomial Regression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
poly_reg= PolynomialFeatures(degree = 4)
X_poly = poly_reg.fit_transform(X_train)
poly_linear_reg = LinearRegression()
poly_linear_reg.fit(X_poly, y_train)
```

```python
# predicting results for poly reg
y_pred_poly = poly_linear_reg.predict(poly_reg.transform(X_test))

r2_score(y_test, y_pred_poly)
```
